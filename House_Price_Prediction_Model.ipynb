# Handling Missing Values using Droping Values

```python
df['Price'].isnull().sum()
```

```python
y = df.Price
```

```python
# Select all object type columns and drop them
X = df.select_dtypes(exclude = 'object')
X.head()
```

```python
X.dtypes
```

```python
# Handling missing values using Drop method
missing_columns =  [ col for col in X.columns if X[col].isnull().any()]
missing_columns.append('Price')
missing_columns
```

```python
reduced_X =  X.drop(missing_columns , axis = 1)
reduced_X.head()
```

# Data Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt



```

```python
from sklearn.model_selection import train_test_split
train_X , test_X ,train_y , test_y = train_test_split(reduced_X,y ,test_size = 0.2 ,random_state = 1)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
```

```python

def get_error(estimator , train_X,test_X,train_y,test_y):
  model = RandomForestRegressor(n_estimators = estimator , random_state = 1)
  model.fit(train_X,train_y)
  prediction = model.predict(test_X)
  return mean_absolute_error(test_y, prediction)
```

```python
my_list = [5,10,15,20,25,30,40,50,60,70,80,90,100]
my_dictionary = {estimator : get_error(estimator,train_X,test_X,train_y,test_y) for estimator  in my_list}
my_dictionary
```

```python
best_tree_num = min(my_dictionary,key = my_dictionary.get)
best_tree_num
```

```python
model = RandomForestRegressor(n_estimators = best_tree_num ,random_state = 1)
model.fit(reduced_X,y)
```
